# 🧠 মেশিন লার্নিং কী এবং কেন?

**স্বাগতম!** আজ আমরা জানবো **মেশিন লার্নিং** (Machine Learning) কী, কেন এটা দরকার, এবং এটা কীভাবে কাজ করে।

তুমি কি জানো? **মেশিন লার্নিং** হলো কম্পিউটারকে শেখানোর একটি পদ্ধতি — ঠিক যেমন তুমি স্কুলে পড়ো এবং শেখো! পৃথিবীর সবচেয়ে স্মার্ট প্রোগ্রামগুলো আজকাল ML ব্যবহার করে কাজ করে। গুগল, ফেসবুক, ইউটিউব, নেটফ্লিক্স — সবাই ML ব্যবহার করে!

এই অধ্যায়ে আমরা ৫টি জিনিস শিখবো:
1. 🧠 **ML কী?** — একটি মজার গল্পের মাধ্যমে
2. 🤔 **কেন ML দরকার?** — বাস্তব জীবনের উদাহরণ
3. 🧩 **ML কী ধরনের সমস্যা সমাধান করে?** — classification, regression, clustering
4. 📚 **তিন ধরণের ML** — supervised, unsupervised, reinforcement
5. 🐍 **পাইথনে Code** — sklearn দিয়ে ডেটা দেখা

**চলো শুরু করি!** 🚀

## 📖 গল্প: কম্পিউটার হলো একজন ছাত্র

### 🍎 আপেল চেনার গল্প

ধরো, তুমি একটি ছোট বাচ্চাকে **আপেল** চিনতে শেখাচ্ছো। তুমি কী করবে?

তুমি বাচ্চাটিকে অনেকগুলি আপেল দেখাবে এবং বলবে:
- "এইটা আপেল" 🍎
- "এইটাও আপেল"
- "এইটা আপেল না, এটা কমলা" 🍊
- "এইটাও আপেল, দেখো লাল আর গোল"

কিছুক্ষণ পর, বাচ্চাটি নিজেই আপেল চিনতে পারবে! সে যদি একটি নতুন লাল ফল দেখে, সাথে সাথে বলবে — "এইটা আপেল!" কিন্তু যদি দেখে হলুদ আর লম্বাটে — বলবে "এইটা আপেল না, এটা কলা!" 🍌

**এটাই মেশিন লার্নিং!** কম্পিউটারকেও আমরা অনেক **উদাহরণ** (examples) দেখাই, আর কম্পিউটার সেগুলো থেকে শেখে। তাকে আলাদা করে কোনো নিয়ম (rules) বানিয়ে দিতে হয় না — সে নিজেই প্যাটার্ন খুঁজে বের করে শিখে নেয়!

### 🎒 আরেকটি গল্প: কীভাবে তুমি বন্ধুকে চিনো?

তোমার ক্লাসে ৪০ জন বন্ধু। তুমি কীভাবে তাদের একে অপরের থেকে আলাদা করে চিনো? তুমি কি করে বুঝো — এই ব্যক্তি রহিম, এই ব্যক্তি করিম?

তুমি তাদের **বৈশিষ্ট্য** (features) দেখে চিনো:
- রহিমের চশমা আছে 👓
- ফাতেমার লম্বা চুল আছে
- করিমের ব্যাগে টম এন্ড জেরির স্টিকার আছে
- আয়েশা সবসময় লাল জামা পরে

ML-ও ঠিক একই কাজ করে! কম্পিউটার প্রতিটি জিনিসের বৈশিষ্ট্য (features) দেখে শেখে — কোন জিনিসটা কী। যেমন একটি ইমেইলের বৈশিষ্ট্য হতে পারে: "কতগুলো বড় হাতের অক্ষর আছে", "কতগুলো ! চিহ্ন আছে", "লটারি শব্দটি আছে কিনা"।

### ⚡ প্রথাগত প্রোগ্রামিং বনাম ML

**প্রথাগত প্রোগ্রামিং:**
```
যদি (রং == লাল) এবং (আকৃতি == গোল) তাহলে:
    উত্তর = "আপেল"
```
→ আমরা **নিয়ম** (rules) দিই

**মেশিন লার্নিং:**
```
১০০০ টি আপেলের ছবি দেখাও → কম্পিউটার নিজেই শেখে → নতুন ছবি দেখলেও চিনতে পারে
```
→ আমরা **উদাহরণ** (examples) দিই

> **মনে রেখো:** নিয়ম দিয়ে প্রোগ্রামিং করলে সব সম্ভাব্যケース আগে থেকে ভাবতে হয়। ML-এ কম্পিউটার নিজেই শিখে নেয় — এমনকি আমরা যা ভাবিনি, সেটাও!

### 🎯 ML শেখার সবচেয়ে মজার উদাহরণ — একটি খেলা

একটি ML গেম খেলো: তুমি ১০টি বিড়াল 🐱 আর ১০টি কুকুর 🐶-এর ছবি দেখাও। কম্পিউটার শেখে বিড়াল আর কুকুরের পার্থক্য। তারপর তুমি একটি নতুন ছবি দেখাও — কম্পিউটার কি বলবে? উত্তর যদি সঠিক হয়, তাহলে কম্পিউটার সত্যিই শিখেছে!

এখন তুমি ML-এর বেসিক আইডিয়া বুঝে গেছো! **ML হলো উদাহরণ থেকে শেখার প্রক্রিয়া।**

## 🤔 কেন মেশিন লার্নিং দরকার?

### প্রথাগত প্রোগ্রামিং-এর সীমাবদ্ধতা

ভাবো তো — তুমি কীভাবে একটি স্প্যাম ইমেইল চিনতে পারো?

"লটারি জিতেছেন!" — এটা নিশ্চয়ই স্প্যাম! কিন্তু স্প্যামাররা সবসময় বুদ্ধি বদলায়:

| পুরনো স্প্যাম | নতুন স্প্যাম | আরও নতুন |
|--------------|-------------|-----------|
| "লটারি জিতেছেন" | "আপনি ১০ লক্ষ টাকা পেয়েছেন" | "কংগ্রাচুলেশনস! ইউ আর আ উইনার!" |
| "বিনামূল্যে আইফোন" | "ফ্রি গিফট কার্ড আপনার জন্য" | "ক্লিক করুন এবং জিতে নিন" |

প্রথাগত প্রোগ্রামিং-এ আমাদের প্রতিটি নতুন শব্দের জন্য নিয়ম বানাতে হতো! কিন্তু স্প্যামাররা হাজার হাজার উপায়ে লেখে — সব নিয়ম বানানো অসম্ভব!

**ML সমাধান:** আমরা কম্পিউটারকে ১০,০০০ স্প্যাম আর ১০,০০০ নন-স্প্যাম ইমেইল দেখাই। কম্পিউটার নিজেই প্যাটার্ন খুঁজে বের করে — "এই ধরনের ইমেইলগুলোর বেশিরভাগই স্প্যাম হয়"। এখন নতুন ইমেইল এলেও সে চিনতে পারে!

### ML কেন এত জনপ্রিয়? তিনটি কারণ

**কারণ ১: অটোমেশন 🤖**
মানুষ ছাড়াই কাজ করতে পারে — যেমন স্প্যাম ফিল্টার দিনরাত কাজ করে, ঘুমায় না, বিশ্রাম নেয় না!

**কারণ ২: প্যাটার্ন খোঁজা 🔍**
মানুষ যা দেখতে পায় না, ML দেখতে পায়। যেমন ML ডাক্তারের চেয়েও সূক্ষ্মভাবে এক্স-রে বিশ্লেষণ করতে পারে। অথবা গুগল ML ব্যবহার করে billions of searches থেকে প্যাটার্ন খুঁজে বের করে!

**কারণ ৩: অভিজ্ঞতা থেকে শেখা 📈**
যত বেশি ডেটা পায়, তত ভালো হয়। যেমন ইউটিউব — তুমি যত বেশি দেখো, তত ভালো suggestion দেয়। প্রথমে খারাপ suggestion দিতে পারে, কিন্তু ধীরে ধীরে শিখে যায় তোমার পছন্দ!

## 🧩 ML-এর সমস্যার ধরন

ML তিন ধরনের সমস্যা সমাধান করে। এগুলো বোঝার জন্য আমরা **ABCD** স্টেপ ব্যবহার করবো।

### 🅰️ টাইপ ১: শ্রেণীবিভাগ (Classification)

**বুঝতে পারো:** তুমি কি মোবাইলে কন্টাক্ট লিস্টে বন্ধুদের গ্রুপ করো? "পরিবার", "বন্ধু", "স্কুল" — একটা নাম দেখে তুমি বলে দিতে পারো কোন গ্রুপে রাখবে? এটাই classification!

**কাজ:** জিনিসকে বিভিন্ন **গ্রুপে** ভাগ করা। উত্তর হয় একটি **শ্রেণী** (category)। যেমন: "এটা কি বিড়াল নাকি কুকুর?"

**ABCD ধাপ:**
> **A** — অনেক ছবি সংগ্রহ করো (যেমন ১০০০ আপেলের ছবি, ১০০০ কমলার ছবি)
> **B** — প্রতিটাতে লেবেল দাও ("এটা আপেল", "এটা কমলা")
> **C** — কম্পিউটারকে এই labeled ডেটা দেখাও
> **D** — কম্পিউটার নিজেই শিখে নেয় — নতুন ছবি দেখলেও চিনতে পারে!

**বাস্তব উদাহরণ:**
- 📧 ইমেইল: স্প্যাম ❌ বা ইনবক্স ✅
- 🏥 রোগ: ক্যান্সার আছে বা নেই
- 🐱 ছবি: বিড়াল 🐱 বা কুকুর 🐶
- ✍️ হাতের লেখা: ০, ১, ২, ..., ৯ কোন সংখ্যা?
- 🍎 ফল: আপেল, কমলা, কলা, আঙ্গুর — কোনটা?

### 🅱️ টাইপ ২: রিগ্রেশন (Regression)

**বুঝতে পারো:** তুমি কি কখনো ভেবেছো "যদি আমি আজ ২ ঘণ্টা পড়ি, তাহলে পরীক্ষায় কত পাব?" — তুমি একটি সংখ্যা অনুমান করছো। এটাই regression!

**কাজ:** একটি **সংখ্যা** অনুমান করা। উত্তর গ্রুপ না, বরং একটা নির্দিষ্ট সংখ্যামান। যেমন: "এই বাড়ির দাম কত হবে?"

**ABCD ধাপ:**
> **A** — পুরনো ডেটা সংগ্রহ করো (যেমন ১০০ বাড়ির আকার, কক্ষ সংখ্যা ও দাম)
> **B** — ডেটাতে বৈশিষ্ট্য (features) চিহ্নিত করো (আকার, কক্ষ সংখ্যা, জায়গার আয়তন)
> **C** — কম্পিউটারকে ডেটা দেখাও — "এই বাড়ির দাম এত"
> **D** — কম্পিউটার নতুন বাড়ির দাম অনুমান করতে পারে!

**বাস্তব উদাহরণ:**
- 🏠 বাড়ি: আকার দেখে দাম কত হবে?
- 🌡️ আবহাওয়া: আজ দেখে কালকের তাপমাত্রা কত হবে?
- 📚 পড়াশোনা: কত ঘণ্টা পড়লে পরীক্ষায় কত নম্বর পাওয়া যাবে?
- 📈 শেয়ার বাজার: আগামীকাল শেয়ারের দাম কত হবে?

> **B** — কোনো লেবেল দিও না (কারণ তুমিও জানো না কে কী গ্রুপে)

> **D** — কম্পিউটার নিজেই গ্রুপ বের করে: "এই গ্রাহকরা বেশি কেনে, ওরা কম কেনে, ওরা মাঝারি"

**বাস্তব উদাহরণ:**

## 📚 তিন ধরণের মেশিন লার্নিং

ML-কে শেখার পদ্ধতি অনুযায়ী তিন ভাগে ভাগ করা হয়। প্রতিটি ভাগ বোঝার জন্য আমরা একটি করে গল্প শিখবো।

### 🅰️ সুপারভাইজড লার্নিং (তত্ত্বাবধানে শিক্ষা) — "শিক্ষক আছে" 👨‍🏫

**গল্প:** ধরো, তুমি একটি ছোট বাচ্চাকে ফলের নাম শেখাচ্ছো। তুমি তার হাতে একটি আপেল দিয়ে বললে "এটা আপেল", তারপর একটি কমলা দিয়ে বললে "এটা কমলা"। তুমি **শিক্ষক** — তুমি *উত্তর* (labels) দিয়ে দিচ্ছো। বাচ্চা শিখছে — "হুম, লাল আর গোল = আপেল, কমলা আর গোল = কমলা"।

**কিভাবে কাজ করে:**
- **Input:** features (বৈশিষ্ট্য) + labels (উত্তর/লেবেল)
- **Goal:** শেখা যে কীভাবে features দেখে label ঠিকঠাক বলা যায়
- **যেমন:** ১০০টি ইমেইল দেওয়া আছে — প্রতিটাতে লেখা আছে "স্প্যাম" বা "ইনবক্স"। কম্পিউটার শেখে কোন ধরনের ইমেইল স্প্যাম হয়।

**উদাহরণ:**
- 💌 স্প্যাম ফিল্টার (ইমেইল + "স্প্যাম" বা "না")
- 🖼️ ছবি চেনা (ছবি + "বিড়াল" বা "কুকুর")
- 🏠 বাড়ির দাম预估 (বাড়ির তথ্য + দাম)

**দুই প্রকার:**
1. **ক্লাসিফিকেশন:** উত্তর হয় একটি **শ্রেণী** (যেমন: বিড়াল/কুকুর)
2. **রিগ্রেশন:** উত্তর হয় একটি **সংখ্যা** (যেমন: ১০ লক্ষ টাকা)

### 🅱️ আনসুপারভাইজড লার্নিং (অনিয়ন্ত্রিত শিক্ষা) — "শিক্ষক নেই" 👤

**গল্প:** এখন তুমি বাচ্চাটিকে বিভিন্ন ফলের একটি ঝুড়ি দিয়ে বললে "এইগুলো সাজাও"। তুমি কিছু বলোনি কোনটা কী। বাচ্চাটি নিজেই দেখে — "এইগুলো লাল আর গোলাকার... এইগুলো হলুদ আর লম্বা... এইগুলো সবুজ আর ছোট" — সে নিজেই গ্রুপ বের করে নেয়! কিন্তু সে জানে না যে "লালগুলো আপেল", সে শুধু জানে "এগুলো একই রকম"।

**কিভাবে কাজ করে:**
- **Input:** শুধু ডেটা (features), কোনো label/উত্তর নেই
- **Goal:** ডেটার মধ্যে প্যাটার্ন বা গ্রুপ খুঁজে বের করা
- **যেমন:** ১০০০ গ্রাহকের কেনাকাটার ডেটা — কম্পিউটার নিজেই খুঁজে বের করে কারা একই ধরনের গ্রাহক

**উদাহরণ:**
- 🛍️ গ্রাহক সেগমেন্টেশন (গ্রাহকদের গ্রুপ)
- 🧬 জিন বিশ্লেষণ (একই ধরনের জিন একসাথে)
- 🎵 মিউজিক রেকমেন্ডেশন (একই ধরনের গান একসাথে)

### 🅲 রিইনফোর্সমেন্ট লার্নিং (শক্তিবর্ধন শিক্ষা) — "চেষ্টা করে শেখো" 🎮

**গল্প:** তুমি একটি নতুন ভিডিও গেম খেলতে শুরু করেছো। প্রথমে তুমি জানো না কী করতে হবে। তুমি চেষ্টা করো:
- ডানে গেলে পয়েন্ট পেলে 🎉
- বামে গেলে গর্তে পড়ে গেম ওভার 💀
- উপরে গেলে একটা তারা পেলে ⭐

ধীরে ধীরে তুমি শিখে যাও — কী করলে বেশি পয়েন্ট পাওয়া যায়, কী করলে গেম ওভার হয়। *এটাই রিইনফোর্সমেন্ট লার্নিং!*

**কিভাবে কাজ করে:**
1. কম্পিউটার একটি **action** নেয় (যেমন: ডানে যাও বা বামে যাও)
2. পরিবেশ থেকে **reward** (পুরস্কার - যেমন পয়েন্ট) বা **penalty** (শাস্তি - যেমন গেম ওভার) পায়
3. কম্পিউটার শেখে — কোন action-এ বেশি reward পাওয়া যায়, কোন action এড়িয়ে চলতে হবে

**উদাহরণ:**
- 🎮 গেম খেলা (AlphaGo, Chess AI, Atari games)
- 🚗 স্বয়ংক্রিয় গাড়ি চালানো (গাড়ি নিজেই শেখে কীভাবে নিরাপদে চালাতে হয়)
- 🤖 রোবট হাঁটা শেখা (রোবট পড়ে যায়, আবার উঠে, ধীরে ধীরে শেখে)
- 🏭 কারখানায় পণ্য সরানো (কোন রুটে পণ্য সরালে সবচেয়ে দ্রুত হয়)

### ⚡ তিন ধরণের ML-এর বিস্তারিত তুলনা

| বৈশিষ্ট্য | Supervised | Unsupervised | Reinforcement |
|-----------|------------|-------------|---------------|
| 📊 **ডেটা** | Labeled (উত্তর + ডেটা) | Unlabeled (শুধু ডেটা) | Trial & Error (Reward) |
| 👨‍🏫 **শিক্ষক** | আছে (উত্তর জানা) | নেই | পরিবেশ (reward) |
| 🎯 **লক্ষ্য** | উত্তর ঠিক করা | প্যাটার্ন খোঁজা | সর্বোচ্চ reward অর্জন |
| 💡 **ব্যবহার** | স্প্যাম, ছবি চেনা | গ্রাহক গ্রুপিং | গেম, রোবট |
| 🔄 **ফিডব্যাক** | প্রতিটি নমুনার জন্য | কোনো ফিডব্যাক নেই | শুধু ভাল/মন্দ |

> ❓ যদি উত্তর দিতে পারো — তাহলে তুমি ML-এর বেসিক পুরোপুরি বুঝে গেছো! 🎉

````python
# 🖥️ কোড টাইম! প্রথমে প্রয়োজনীয় লাইব্রেরি ইম্পোর্ট করি
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
print("📦 সব লাইব্রেরি ইম্পোর্ট হয়েছে!")
print()

# scikit-learn থেকে একটি ডেটাসেট লোড করি
# digits ডেটাসেট: হাতের লেখা 0-9 পর্যন্ত সংখ্যার ছবি
# এটা supervised classification-এর উদাহরণ (ছবি + সংখ্যা লেবেল)
digits = datasets.load_digits()
print("🔢 **ডিজিটস (Digits) ডেটাসেট**")
print("=" * 40)
print("এটা 8x8 পিক্সেলের ছোট ছবি — হাতে লেখা সংখ্যা")
print()
print(f"📐 ডেটাসেটের আকার (shape): {digits.data.shape}")
print(f"   → {digits.data.shape[0]} টি ছবি, প্রতিটি {digits.data.shape[1]} টি পিক্সেল")
print(f"📐 ছবির আকার (images shape): {digits.images.shape}")
print(f"   → {digits.images.shape[0]} টি ছবি, প্রতিটি {digits.images.shape[1]}x{digits.images.shape[2]} পিক্সেল")
print(f"\n🏷️ লক্ষ্য (target) নামগুলো: {digits.target_names}")
print(f"🏷️ প্রথম 20টি লক্ষ্যের মান: {digits.target[:20]}")
print()
print("💡 মনে রেখো: এখানে features = পিক্সেলের মান (8x8=64টি)")
print("          target = সংখ্যাটি কী (0-9)")
````

````python
# এখন কিছু ছবি দেখি — কম্পিউটার কীভাবে ডেটা "দেখে"
plt.figure(figsize=(14, 6))

for i in range(16):
    ax = plt.subplot(4, 4, i + 1)
    ax.imshow(digits.images[i], cmap='gray')
    ax.set_title(f"এটা {digits.target[i]}", fontsize=12, fontweight='bold')
    ax.axis('off')

plt.suptitle("🖼️ হাতের লেখা সংখ্যা — ডেটাসেটের প্রথম 16টি ছবি", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
print("👆 উপরের ছবিগুলো দেখো! প্রতিটি 8x8 পিক্সেলের ছোট ছবি।")
print("কম্পিউটার এই ছবির পিক্সেলের মান দেখে শেখে — কোন আকৃতি কোন সংখ্যা।")
print("এটা supervised classification-এর উদাহরণ!")
````

## 🔬 আরেকটি উদাহরণ: ইরিস (Iris) ফুলের ডেটাসেট 🌸

এখন আমরা আরেকটি বিখ্যাত ডেটাসেট দেখবো — **ইরিস ফুল**। এই ডেটাসেট ১৯৩৬ সাল থেকে ব্যবহৃত হচ্ছে! 😮 বিজ্ঞানী রোনাল্ড ফিশার এই ডেটাসেট তৈরি করেছিলেন এবং এখনো ML শেখার জন্য এটাই সবচেয়ে জনপ্রিয় প্রথম ডেটাসেট!

**কি আছে এই ডেটাসেটে?**
- ৩ প্রজাতির ফুল: Setosa, Versicolor, Virginica 🌸
- মোট ১৫০টি নমুনা (প্রতি প্রজাতিতে ৫০টি করে)
- ৪টি বৈশিষ্ট্য (features):
  1. 📏 সেপালের দৈর্ঘ্য (sepal length) — cm-এ
  2. 📐 সেপালের প্রস্থ (sepal width) — cm-এ
  3. 📏 পাপড়ির দৈর্ঘ্য (petal length) — cm-এ
  4. 📐 পাপড়ির প্রস্থ (petal width) — cm-এ

কম্পিউটার এসব বৈশিষ্ট্য দেখে শেখে — কোন ফুল কোন প্রজাতির!

````python
# ইরিস ডেটাসেট লোড করি
iris = datasets.load_iris()
print("🌸 **ইরিস (Iris) ডেটাসেট**")
print("=" * 40)
print()
print(f"📋 ফিচারের নাম: {iris.feature_names}")
print(f"🏷️ লক্ষ্যের নাম (ফুলের প্রজাতি): {iris.target_names}")
print()
print(f"📐 ডেটার আকার (samples x features): {iris.data.shape}")
print(f"   → {iris.data.shape[0]} টি ফুল, প্রতিটির {iris.data.shape[1]} টি বৈশিষ্ট্য")
print()
print("📄 প্রথম 5টি নমুনা (features):")
print(iris.data[:5])
print()
print("🏷️ প্রথম 5টি লক্ষ্য (0=Setosa, 1=Versicolor, 2=Virginica):")
print(iris.target[:5])
````

````python
# ইরিস ডেটা ভিজুয়ালাইজ করি — স্ক্যাটার প্লট
# X = পাপড়ির দৈর্ঘ্য (index 2), Y = পাপড়ির প্রস্থ (index 3)
x_index = 2  # petal length (cm)
y_index = 3  # petal width (cm)

# তিন রঙে তিন প্রজাতির ফুল দেখাই
colors = ['red', 'blue', 'green']
markers = ['o', 's', '^']

plt.figure(figsize=(10, 7))

for i, (color, marker) in enumerate(zip(colors, markers)):
    subset = iris.target == i
    plt.scatter(
        iris.data[subset, x_index],
        iris.data[subset, y_index],
        c=color, marker=marker, label=iris.target_names[i],
        alpha=0.8, s=80, edgecolors='black', linewidth=0.5
    )

plt.xlabel(iris.feature_names[x_index] + " (cm)", fontsize=12)
plt.ylabel(iris.feature_names[y_index] + " (cm)", fontsize=12)
plt.title("🌸 ইরিস ফুল: পাপড়ির দৈর্ঘ্য vs প্রস্থ\n(তিন প্রজাতির ফুল আলাদা দেখা যাচ্ছে)", fontsize=14)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.show()
print("👆 দেখো, তিন রঙের ফুলগুলো আলাদা জায়গায় আছে!")
print("লাল (Setosa) পুরোপুরি আলাদা — নিচে বামে।")
print("নীল (Versicolor) মাঝখানে, সবুজ (Virginica) উপরে ডানে।")
print("কম্পিউটার এই প্যাটার্ন দেখে শিখে ফুল চিনতে পারে।")
print()
print("💡 **মনে রেখো:** এই প্লট দেখায় কেন ML কাজ করে!")
print("যদি তুমিও এই প্যাটার্ন দেখো, তুমিও ফুল চিনতে পারবে।")
````

## 🎯 ML পাইপলাইন — একটি প্রজেক্টের ধাপ

প্রথমবার ML প্রজেক্ট করলে মনে হতে পারে "কোথায় শুরু করবো?"। চিন্তা নেই! আমরা ৬টি সহজ ধাপে ML করি:

**ABCDEF — ML প্রজেক্টের ৬ ধাপ:**

| ধাপ | নাম (ইংরেজি) | কী করি |
|-----|-------------|--------|
| **A** | **A**cquire | ডেটা সংগ্রহ/লোড করা (যেমন sklearn থেকে) |
| **B** | **B**rowse | ডেটা দেখা, বৈশিষ্ট্য (features) বোঝা |
| **C** | **C**lean | ডেটা পরিষ্কার করা (ভুল/নিখোঁজ ডেটা সরানো) |
| **D** | **D**ivide | ডেটা ভাগ করা: train (শেখার জন্য) + test (পরীক্ষার জন্য) |
| **E** | **E**stimate | ML মডেল train/shikhao (fit) |
| **F** | **F**ind | মূল্যায়ন — accuracy দেখা, মডেল কতটা ভালো কাজ করে |

আজকের শেষ দুই কোড সেলে আমরা ধাপ A আর B করেছি (ডেটা লোড + দেখা)। পরবর্তী পাঠে আমরা C, D, E, F করবো!

এখন পর্যন্ত আমরা শিখেছি ML কী, কেন দরকার, কত প্রকার — পরবর্তী কোডগুলোতে বাস্তব ML দেখা শুরু করবো।
