**ধাপ ১**: রুমা, যে ক্যাপ্টেন হতে চায়, সামনে এগিয়ে গিয়ে বলল, "যারা আমার সাথে নাচতে চাও, আমার চারপাশে এসো!"

**ধাপ ৩**: প্রতিটি দলের সদস্য দেখল তাদের ক্যাপ্টেন কে। কিন্তু মনে হলো কিছু সদস্য মাঝখানে দাঁড়িয়ে — তারা নিশ্চিত না কার দলে যাবে।

**ধাপ ৪**: তাই প্রতিটি দল আবার ভোট দিয়ে নতুন ক্যাপ্টেন বেছে নিল — যে দলের সবচেয়ে মাঝখানে আছে।

**ধাপ ৫**: আবার সবাই নতুন ক্যাপ্টেনের চারপাশে জড়ো হলো।

**ধাপ ৬**: এই প্রক্রিয়া বারবার চলতে লাগলো — যতক্ষণ না সবাই নিশ্চিত যে তারা সঠিক দলে আছে।

অবশেষে, সবাই কিছু দলে বিভক্ত হয়ে গেল। প্রতিটি দলের নিজস্ব ক্যাপ্টেন এবং নিজস্ব জায়গা।

এই পুরো প্রক্রিয়াটাই হলো **K-Means ক্লাস্টারিং**! এখানে:
- **K** = কতগুলো দল (গ্রুপ) আমরা চাই
- **Means** = প্রতিটি দলের "ক্যাপ্টেন" আসলে দলের সব পয়েন্টের গড় (mean) অবস্থান
- ডেটা পয়েন্টগুলো নিজেদের মধ্যে দূরত্ব মেপে গ্রুপ তৈরি করে

K-Means হল একটি **অতত্ত্বাবধানে শিক্ষণ** পদ্ধতি — কারণ এটা আমাদের ডেটার লেবেল (উত্তর) দেখে না, শুধু ডেটার প্যাটার্ন দেখে নিজে থেকেই গ্রুপ তৈরি করে।

## 🎯 K-Means অ্যালগরিদম (ধাপে ধাপে)

K-Means অ্যালগরিদমটি ৪টি ধাপে কাজ করে:

### 🔵 ধাপ ১: K সংখ্যক সেন্ট্রয়েড বেছে নেওয়া
প্রথমে আমরা এলোমেলোভাবে K সংখ্যক পয়েন্ট বেছে নেই — এগুলো হবে প্রাথমিক সেন্ট্রয়েড (ক্লাস্টারের কেন্দ্র)।

### 🟢 ধাপ ২: পয়েন্ট অ্যাসাইনমেন্ট
প্রতিটি ডেটা পয়েন্টকে সবচেয়ে কাছের সেন্ট্রয়েডের সাথে যুক্ত করা হয়। এভাবেই Kটি ক্লাস্টার তৈরি হয়।

### 🟡 ধাপ ৩: সেন্ট্রয়েড আপডেট
প্রতিটি ক্লাস্টারের জন্য নতুন সেন্ট্রয়েড গণনা করা হয় — এটি ঐ ক্লাস্টারের সব পয়েন্টের গড় (mean) অবস্থান।

### 🔴 ধাপ ৪: পুনরাবৃত্তি
ধাপ ২ এবং ৩ বারবার করা হয় যতক্ষণ না সেন্ট্রয়েডের অবস্থান পরিবর্তন হওয়া বন্ধ করে (কনভার্জ)।

### 📝 গুরুত্বপূর্ণ প্যারামিটার:
- **n_clusters (K)** = আমরা কতগুলো গ্রুপ চাই
- **random_state** = এলোমেলো সংখ্যা নিয়ন্ত্রণ (একই ফলাফল পেতে)
- **n_init** = কতবার ভিন্ন ভিন্ন প্রাথমিক সেন্ট্রয়েড দিয়ে চেষ্টা করবে (ডিফল্ট ১০)
- **max_iter** = সর্বোচ্চ কত ধাপ পর্যন্ত যাবে

### ⚠️ K-Means-এর সীমাবদ্ধতা:
- আমাদের আগে থেকে K-র মান বলতে হবে
- গোলাকার ক্লাস্টারের জন্য ভালো কাজ করে, অদ্ভুত আকৃতির জন্য না
- আউটলায়ারের প্রতি সংবেদনশীল
- এলোমেলো শুরুর কারণে প্রতি বার ভিন্ন ফল আসতে পারে

```python
# প্রয়োজনীয় লাইব্রেরি
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs, make_moons, load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
print("✅ সব লাইব্রেরি লোড হয়েছে!")
```

## 🗂️ আমাদের ডেটা তৈরি

আমরা make_blobs ব্যবহার করে কিছু কৃত্রিম ডেটা তৈরি করব। Blobs মানে হচ্ছে ডেটার পাহাড়ের মতো গ্রুপ।

```python
# কৃত্রিম ডেটা তৈরি
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
print("📊 ডেটার আকৃতি:", X.shape)
print("🎯 আসল ক্লাস্টার: 4টি")
print("📝 ডেটা পয়েন্ট: 300টি")

# ডেটা প্লট
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y_true, cmap='viridis', edgecolors='k', s=60)
plt.title('আমাদের ডেটা (আসল ক্লাস্টার)', fontsize=14, fontweight='bold')
plt.xlabel('ফিচার 1', fontsize=12)
plt.ylabel('ফিচার 2', fontsize=12)
plt.show()
print("👆 আমরা পরিষ্কারভাবে 4টি গ্রুপ দেখতে পাচ্ছি!")
```

## 🧪 K-Means চালানো

এখন আমরা K-Means অ্যালগরিদম চালাব এবং দেখব এটি কতটা ভালোভাবে গ্রুপ তৈরি করতে পারে।

```python
# K-Means ক্লাস্টারিং (K=4)
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X)
print("✅ K-Means সম্পন্ন হয়েছে!")
print(f"🔢 ক্লাস্টার লেবেল:\n{y_kmeans[:20]}")
print(f"📍 সেন্ট্রয়েড:\n{kmeans.cluster_centers_}")
```

```python
# K-Means ফলাফল ভিজুয়ালাইজেশন
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# আসল ক্লাস্টার
axes[0].scatter(X[:, 0], X[:, 1], c=y_true, cmap='viridis', edgecolors='k', s=60)
axes[0].set_title('আসল ক্লাস্টার', fontsize=14, fontweight='bold')
axes[0].set_xlabel('ফিচার 1', fontsize=12)
axes[0].set_ylabel('ফিচার 2', fontsize=12)

# K-Means ক্লাস্টার
axes[1].scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap='viridis', edgecolors='k', s=60)
axes[1].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
                c='red', marker='X', s=200, linewidths=3, edgecolors='black', 
                label='সেন্ট্রয়েড')
axes[1].set_title('K-Means ক্লাস্টার', fontsize=14, fontweight='bold')
axes[1].set_xlabel('ফিচার 1', fontsize=12)
axes[1].set_ylabel('ফিচার 2', fontsize=12)
axes[1].legend()

plt.tight_layout()
plt.show()
print("👆 K-Means প্রায় আসল ক্লাস্টারের মতোই গ্রুপ তৈরি করতে পেরেছে!")
```

## 🔄 K-Means স্টেপ বাই স্টেপ দেখা

চলো, আমরা K-Means-এর প্রতিটি ধাপ ভিজুয়ালি দেখি — কিভাবে সেন্ট্রয়েডগুলো gradually সঠিক জায়গায় চলে আসে।

```python
# K-Means-এর ধাপগুলো দেখা
# আমরা নিজেরাই K-Means এর মত একটি অ্যালগরিদম লিখব

def kmeans_steps(X, k, max_steps=6):
    np.random.seed(42)
    idx = np.random.choice(X.shape[0], k, replace=False)
    centroids = X[idx].copy()
    
    fig, axes = plt.subplots(2, 3, figsize=(16, 10))
    axes = axes.ravel()
    
    for step in range(max_steps):
        # Assign points to nearest centroid
        distances = np.sqrt(((X - centroids[:, np.newaxis])**2).sum(axis=2))
        labels = np.argmin(distances, axis=0)
        
        # Plot
        axes[step].scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', 
                          edgecolors='k', s=50, alpha=0.7)
        axes[step].scatter(centroids[:, 0], centroids[:, 1], 
                          c='red', marker='X', s=200, linewidths=3, edgecolors='black')
        axes[step].set_title(f'ধাপ {step+1}', fontsize=14, fontweight='bold')
        
        # Update centroids
        new_centroids = np.array([X[labels == i].mean(axis=0) for i in range(k)])
        
        # Draw arrows showing centroid movement
        for old, new in zip(centroids, new_centroids):
            if not np.allclose(old, new):
                axes[step].arrow(old[0], old[1], new[0]-old[0], new[1]-old[1],
                                head_width=0.1, head_length=0.1, fc='yellow', ec='yellow', alpha=0.8)
        
        centroids = new_centroids
    
    plt.tight_layout()
    plt.show()
    return centroids, labels

# 3টি ক্লাস্টার নিয়ে দেখি
X_small, _ = make_blobs(n_samples=150, centers=3, cluster_std=0.8, random_state=0)
final_centroids, final_labels = kmeans_steps(X_small, 3)
print("👆 প্রতিটি ধাপে সেন্ট্রয়েড (লাল X) কিভাবে সঠিক জায়গায় চলে আসছে দেখো!")
```

## 📏 Inertia এবং Elbow Method

### Inertia কী?
Inertia হলো প্রতিটি পয়েন্ট এবং তার নিজের সেন্ট্রয়েডের মধ্যে দূরত্বের বর্গের যোগফল।
- **ছোট Inertia** = পয়েন্টগুলো তাদের সেন্ট্রয়েডের কাছাকাছি (ভালো ক্লাস্টারিং)
- **বড় Inertia** = পয়েন্টগুলো সেন্ট্রয়েড থেকে দূরে (খারাপ ক্লাস্টারিং)

### Elbow Method কী?
আমরা বিভিন্ন K-র মানের জন্য K-Means চালাই এবং Inertia গণনা করি। তারপর K বনাম Inertia-র গ্রাফ আঁকি।

গ্রাফে একটা বাঁক (elbow) থাকে — যেখানে Inertia কমার হার হঠাৎ কমে যায়। এই বাঁক বিন্দুটি হলো সেরা K-র মান!

### 🧪 উদাহরণ:
ভাবো, তুমি বাজারে ফল কিনছ। ১ কেজি ফলে কত টাকা? ১০০টি ফলে কত টাকা? প্রথম কয়েক কেজির জন্য দাম বেশি, কিন্তু তারপর দাম কমতে থাকে। একটা সময় পরে দাম খুব ধীরে কমে — সেটাই হলো elbow!

চলো, আমরা আমাদের ডেটার জন্য elbow plot তৈরি করি।

```python
# Elbow Method
inertias = []
K_range = range(1, 11)

for k in K_range:
    kmeans_k = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_k.fit(X)
    inertias.append(kmeans_k.inertia_)
print(f"K = {k}: Inertia = {kmeans_k.inertia_:.2f}")

# Elbow plot
plt.figure(figsize=(10, 6))
plt.plot(K_range, inertias, 'bo-', markersize=10, linewidth=3)
plt.xlabel('ক্লাস্টারের সংখ্যা (K)', fontsize=14)
plt.ylabel('Inertia', fontsize=14)
plt.title('Elbow Method: K vs Inertia', fontsize=16, fontweight='bold')
plt.xticks(K_range)
plt.grid(True, alpha=0.3)

# Elbow point highlight
plt.axvline(x=4, color='red', linestyle='--', alpha=0.7, label='Elbow at K=4')
plt.legend()
plt.show()
print("👆 K=4-এর পর Inertia কমার হার কমে গেছে! তাই K=4 হচ্ছে সেরা মান।")
```

## 🔬 বিভিন্ন K-র সাথে ক্লাস্টারিং

চলো, বিভিন্ন K-র মান দিয়ে K-Means চালাই এবং দেখি ক্লাস্টারগুলো কেমন হয়।

```python
# বিভিন্ন K-র মানের জন্য ক্লাস্টারিং ফলাফল
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

for i, k in enumerate([2, 3, 4, 5, 6, 7]):
    row = i // 3
    col = i % 3
    
    kmeans_k = KMeans(n_clusters=k, random_state=42, n_init=10)
    y_k = kmeans_k.fit_predict(X)
    
    axes[row, col].scatter(X[:, 0], X[:, 1], c=y_k, cmap='viridis', 
                          edgecolors='k', s=50, alpha=0.7)
    axes[row, col].scatter(kmeans_k.cluster_centers_[:, 0], kmeans_k.cluster_centers_[:, 1],
                          c='red', marker='X', s=150, linewidths=2, edgecolors='black')
    axes[row, col].set_title(f'K = {k} (Inertia = {kmeans_k.inertia_:.0f})', 
                            fontsize=13, fontweight='bold')
    axes[row, col].set_xlabel('ফিচার 1', fontsize=10)
    axes[row, col].set_ylabel('ফিচার 2', fontsize=10)

plt.tight_layout()
plt.show()
print("👆 K বাড়ার সাথে সাথে ক্লাস্টারগুলো ছোট হতে থাকে এবং Inertia কমতে থাকে।")
print("K=4 আমাদের জন্য সেরা — এটা আসল ক্লাস্টার সংখ্যার সাথে মিলে যায়!")
```

## 🌟 Iris ডেটাতে K-Means

এবার আমরা পরিচিত Iris ডেটাসেটে K-Means প্রয়োগ করি এবং দেখি অ্যালগরিদম আসল ক্লাস্টার (Setosa, Versicolor, Virginica) কতটা ভালোভাবে খুঁজে পায়।

```python
# Iris ডেটাতে K-Means
iris = load_iris()
X_iris = iris.data
y_iris = iris.target

scaler = StandardScaler()
X_iris_scaled = scaler.fit_transform(X_iris)

kmeans_iris = KMeans(n_clusters=3, random_state=42, n_init=10)
y_iris_kmeans = kmeans_iris.fit_predict(X_iris_scaled)

# ফলাফল দেখা
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

colors_iris = ['red', 'green', 'blue']

# আসল ক্লাস
for i, name in enumerate(iris.target_names):
    axes[0].scatter(X_iris_scaled[y_iris==i, 0], X_iris_scaled[y_iris==i, 1],
                   c=colors_iris[i], label=name, edgecolors='k', s=60, alpha=0.8)
axes[0].set_title('আসল ক্লাস (Iris)', fontsize=14, fontweight='bold')
axes[0].legend()

# K-Means ক্লাস্টার
for i in range(3):
    axes[1].scatter(X_iris_scaled[y_iris_kmeans==i, 0], X_iris_scaled[y_iris_kmeans==i, 1],
                   c=colors_iris[i], edgecolors='k', s=60, alpha=0.8)
axes[1].scatter(kmeans_iris.cluster_centers_[:, 0], kmeans_iris.cluster_centers_[:, 1],
               c='black', marker='X', s=200, linewidths=3, label='সেন্ট্রয়েড')
axes[1].set_title('K-Means ক্লাস্টার (Iris)', fontsize=14, fontweight='bold')
axes[1].legend()

for ax in axes:
    ax.set_xlabel('ফিচার 1', fontsize=11)
    ax.set_ylabel('ফিচার 2', fontsize=11)

plt.tight_layout()
plt.show()
print("👆 K-Means Iris ডেটার 3টি ক্লাস (প্রজাতি) মোটামুটি ভালোভাবে আলাদা করতে পেরেছে!")
```

## 🔍 K-Means-এর সীমাবদ্ধতা

K-Means সবসময় ভালো কাজ করে না। বিশেষ করে যখন:
1. **ক্লাস্টারের আকৃতি গোলাকার না হলে** — উদাহরণস্বরূপ চাঁদের আকৃতির (moon-shaped) ডেটা
2. **ক্লাস্টারের সাইজ খুব আলাদা হলে** — কিছু বড়, কিছু ছোট
3. **আউটলায়ার থাকলে** — দূরের পয়েন্টগুলো সেন্ট্রয়েডের অবস্থান নষ্ট করে দেয়

চলো, আমরা make_moons দিয়ে এমন ডেটা তৈরি করি যেখানে K-Means ভালো কাজ করবে না।

```python
# K-Means-এর সীমাবদ্ধতা দেখা
X_moons, y_moons = make_moons(n_samples=300, noise=0.05, random_state=0)

kmeans_moons = KMeans(n_clusters=2, random_state=42, n_init=10)
y_moons_kmeans = kmeans_moons.fit_predict(X_moons)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# আসল আকৃতি
axes[0].scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons, cmap='viridis',
               edgecolors='k', s=60)
axes[0].set_title('আসল ডেটা (2টি চাঁদের আকৃতি)', fontsize=14, fontweight='bold')
axes[0].set_xlabel('ফিচার 1', fontsize=12)
axes[0].set_ylabel('ফিচার 2', fontsize=12)

# K-Means ফলাফল
axes[1].scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons_kmeans, cmap='viridis',
               edgecolors='k', s=60)
axes[1].scatter(kmeans_moons.cluster_centers_[:, 0], kmeans_moons.cluster_centers_[:, 1],
               c='red', marker='X', s=200, linewidths=3, edgecolors='black', label='সেন্ট্রয়েড')
axes[1].set_title('K-Means ফলাফল (ভুল!)', fontsize=14, fontweight='bold')
axes[1].set_xlabel('ফিচার 1', fontsize=12)
axes[1].set_ylabel('ফিচার 2', fontsize=12)
axes[1].legend()

plt.tight_layout()
plt.show()
print("👆 K-Means চাঁদের আকৃতির ডেটা সঠিকভাবে আলাদা করতে পারেনি!")
print("এ কারণে আমাদের DBSCAN-এর মতো অন্য অ্যালগরিদম দরকার হয়।")
print("সেটা আমরা পরবর্তী পাঠে শিখব।")
```

## 🎮 খেলা: নিজে করো!

এখন তুমি বিভিন্ন K-র মান, বিভিন্ন ডেটা নিয়ে পরীক্ষা করো।

```python
# 🎯 তুমি এখানে নিজে এক্সপেরিমেন্ট করো!

my_k = 5  # K-র মান পরিবর্তন করো
my_data_n = 400  # ডেটা পয়েন্ট সংখ্যা
my_centers = 5  # আসল ক্লাস্টার সংখ্যা

# নতুন ডেটা তৈরি
X_new, y_new = make_blobs(n_samples=my_data_n, centers=my_centers, 
                          cluster_std=0.7, random_state=42)

kmeans_my = KMeans(n_clusters=my_k, random_state=42, n_init=10)
y_my = kmeans_my.fit_predict(X_new)

plt.figure(figsize=(8, 6))
plt.scatter(X_new[:, 0], X_new[:, 1], c=y_my, cmap='viridis', edgecolors='k', s=50)
plt.scatter(kmeans_my.cluster_centers_[:, 0], kmeans_my.cluster_centers_[:, 1],
           c='red', marker='X', s=150, linewidths=2, edgecolors='black')
plt.title(f'K-Means: K = {my_k}', fontsize=14, fontweight='bold')
plt.xlabel('ফিচার 1', fontsize=12)
plt.ylabel('ফিচার 2', fontsize=12)
plt.show()
print(f"📊 Inertia: {kmeans_my.inertia_:.2f}")
print()
print("💡 উপরের কোডে my_k, my_data_n, my_centers পরিবর্তন করে দেখো কী হয়!")
```

## 🌟 মজার তথ্য!

🎯 **K-Means** ১৯৫৭ সালে James MacQueen প্রথম প্রস্তাব করেন। এটি এখনও সবচেয়ে জনপ্রিয় ক্লাস্টারিং অ্যালগরিদম!

🎯 **K** (ক্লাস্টার সংখ্যা) বেছে নেওয়ার জন্য শুধু Elbow Method নয়, **Silhouette Score**-ও ব্যবহার করা যায় (পরবর্তী পাঠে শিখব)।

🎯 **K-Means++** হলো K-Means-এর একটি উন্নত সংস্করণ — এটি প্রাথমিক সেন্ট্রয়েড বুদ্ধিমানের সাথে বেছে নেয় (scikit-learn-এ n_init-এর সাথে ব্যবহৃত হয়)।

🎯 **K-Means** আসলে **Vector Quantization**-এর একটি বিশেষ রূপ — এটি ডেটাকে Kটি representative পয়েন্টে সারসংক্ষেপ করে।

🎯 K-Means ইমেজ কম্প্রেশন-এ ব্যবহার করা হয়! প্রতিটি পিক্সেলের রঙকে তার nearest cluster center দিয়ে প্রতিস্থাপন করে ছবির সাইজ কমানো যায়।
