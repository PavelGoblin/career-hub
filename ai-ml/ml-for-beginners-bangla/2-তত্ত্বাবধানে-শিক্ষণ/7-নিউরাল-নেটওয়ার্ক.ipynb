# নিউরাল নেটওয়ার্ক (Neural Network)**গল্প: মস্তিষ্কের মতো গণনা**আমাদের মস্তিষ্কে কোটি কোটি নিউরন (neurons) আছে। প্রতিটি নিউরন কিছু তথ্য পায়, সেটা প্রক্রিয়া করে, এবং পরবর্তী নিউরনে পাঠায়।কৃত্রিম নিউরাল নেটওয়ার্ক (Artificial Neural Network) ঠিক এভাবেই কাজ করে! এটি অনেকগুলো 'নিউরন'-এর স্তর (layer) দিয়ে তৈরি, যেখানে প্রতিটি স্তর কিছু বৈশিষ্ট্য শেখে এবং পরবর্তী স্তরে পাঠায়।

## আমরা কী শিখব?1. MLPClassifier (Multi-Layer Perceptron)2. hidden_layer_sizes (লুকানো স্তরের আকার)3. activation ফাংশন (relu, tanh, logistic)4. বিভিন্ন আর্কিটেকচারের তুলনা

## প্রথমে লাইব্রেরি ইম্পোর্ট করি

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import load_iris, make_classification
from sklearn.model_selection import train_test_split
print('লাইব্রেরি ইম্পোর্ট সফল হয়েছে!')
```

## Iris ডেটাসেট লোড করি

```python
iris = load_iris()
X = iris.data
y = iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42)
print('ট্রেন সেট:', X_train.shape[0], 'নমুনা')
print('টেস্ট সেট:', X_test.shape[0], 'নমুনা')
```

## MLPClassifier তৈরিএকটি সরল নিউরাল নেটওয়ার্ক তৈরি করি: 1টি hidden layer যাতে 10টি নিউরন আছে।

```python
mlp = MLPClassifier(hidden_layer_sizes=(10,),
                    activation='relu', max_iter=1000, random_state=42)
mlp.fit(X_train, y_train)
print('ট্রেন accuracy:', mlp.score(X_train, y_train))
print('টেস্ট accuracy:', mlp.score(X_test, y_test))
```

## hidden_layer_sizes-এর প্রভাবhidden_layer_sizes নির্ধারণ করে নেটওয়ার্কের জটিলতা।- (5,): 1টি layer, 5টি নিউরন (সরল)- (50,): 1টি layer, 50টি নিউরন (জটিল)- (10, 5): 2টি layer (গভীর)নিচে বিভিন্ন আকার试试 করি।

```python
architectures = [
    ('(5,)', (5,)),
    ('(10,)', (10,)),
    ('(50,)', (50,)),
    ('(100,)', (100,)),
    ('(10, 5)', (10, 5)),
    ('(20, 10)', (20, 10))
]
names = []
train_accs = []
test_accs = []
for name, arch in architectures:
    mlp = MLPClassifier(hidden_layer_sizes=arch,
                        activation='relu', max_iter=1000, random_state=42)
    mlp.fit(X_train, y_train)
    names.append(name)
    train_accs.append(mlp.score(X_train, y_train))
    test_accs.append(mlp.score(X_test, y_test))
x = np.arange(len(names))
width = 0.35
plt.bar(x - width/2, train_accs, width, label='Train')
plt.bar(x + width/2, test_accs, width, label='Test')
plt.xticks(x, names, rotation=45)
plt.ylabel('Accuracy')
plt.legend()
plt.title('Architecture vs Accuracy')
plt.show()
```

## activation ফাংশনের তুলনাactivation ফাংশন নির্ধারণ করে একটি নিউরন কীভাবে ইনপুট প্রক্রিয়া করবে। সাধারণ activation ফাংশন:- **relu**: Rectified Linear Unit (সবচেয়ে জনপ্রিয়)- **tanh**: Hyperbolic Tangent (-1 থেকে 1-এর মধ্যে)- **logistic**: Sigmoid (0 থেকে 1-এর মধ্যে)নিচে তিনটি activation ফাংশন তুলনা করি।

```python
activations = ['relu', 'tanh', 'logistic']
x = np.arange(len(activations))
width = 0.35
train_scores = []
test_scores = []
for act in activations:
    mlp = MLPClassifier(hidden_layer_sizes=(10,),
                        activation=act, max_iter=1000, random_state=42)
    mlp.fit(X_train, y_train)
    train_scores.append(mlp.score(X_train, y_train))
    test_scores.append(mlp.score(X_test, y_test))
plt.bar(x - width/2, train_scores, width, label='Train')
plt.bar(x + width/2, test_scores, width, label='Test')
plt.xticks(x, activations)
plt.ylabel('Accuracy')
plt.legend()
plt.title('Activation Function vs Accuracy')
plt.show()
```

## Decision Boundary: বিভিন্ন আর্কিটেকচার2টি feature দিয়ে বিভিন্ন নিউরাল নেটওয়ার্ক আর্কিটেকচারের decision boundary দেখি।

```python
X2, y2 = make_classification(n_features=2, n_redundant=0,
                             n_informative=2, random_state=42)
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, random_state=42)
archs = [
    ('(3,)', (3,)),
    ('(10,)', (10,)),
    ('(50,)', (50,)),
    ('(10, 5)', (10, 5))
]
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.ravel()
for i, (name, arch) in enumerate(archs):
    mlp = MLPClassifier(hidden_layer_sizes=arch,
                        activation='relu', max_iter=1000, random_state=42)
    mlp.fit(X2_train, y2_train)
    x_min, x_max = X2[:, 0].min() - 1, X2[:, 0].max() + 1
    y_min, y_max = X2[:, 1].min() - 1, X2[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.05),
                         np.arange(y_min, y_max, 0.05))
    Z = mlp.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    axes[i].contourf(xx, yy, Z, alpha=0.3)
    axes[i].scatter(X2_train[:, 0], X2_train[:, 1],
                    c=y2_train, edgecolors='k')
    axes[i].set_title('{} acc={:.3f}'.format(name, mlp.score(X2_test, y2_test)))
plt.tight_layout()
plt.show()
```

## Learning Curve (Convergence)MLPClassifier প্রশিক্ষণের সময় loss কমতে থাকে। loss_curve অ্যাট্রিবিউটে এটি দেখা যায়।

```python
mlp = MLPClassifier(hidden_layer_sizes=(10, 5),
                    activation='relu', max_iter=500, random_state=42)
mlp.fit(X_train, y_train)
plt.plot(mlp.loss_curve_)
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Loss Curve')
plt.grid(True)
plt.show()
print('শেষ loss:', mlp.loss_curve_[-1])
```

|------------|----------|

| activation | নিউরনের সক্রিয়করণ ফাংশন |
| max_iter | প্রশিক্ষণের পুনরাবৃত্তি সংখ্যা |
